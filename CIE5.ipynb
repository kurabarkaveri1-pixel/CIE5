```python
from nltk.util import ngrams
from nltk.tokenize import wordpunct_tokenize
```

```python
sentence="""AI is simulation of human intelligence by machine.It includes expects system NLP speech recognition and machine vision."""
```

```python
words=wordpunct_tokenize(sentence)
print("words:")
print(words)
```

```python
bigrams=list(ngrams(words,2))
print("bigram:")
for bigram in bigrams:
  print(bigram)
```

```python
trigrams=list(ngrams(words,2))
print("trigram:")
for trigram in bigrams:
  print(trigram)
```
